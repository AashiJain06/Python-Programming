```python
#Conditional Statement
marks = 85

if marks >= 90:
    print("A+")
elif marks >= 80:
    print("A")
elif marks >= 70:
    print("B")
elif marks >= 60:
    print("C")
else:
    print("Fail")

#This is also allowed in python

if age >= 18 and age <= 60:
    print("Eligible")
if 18 <= age <= 60:
  print("Chained Comparison")

age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("Not eligible")

#Ternary Operator in python
result = "Adult" if age >= 18 else "Minor"
print(result)
```
